# Data Cleaning

The dataset was checked for missing values and duplicate records.

Missing numerical values were filled using the median, while categorical missing values were filled using the mode. Duplicate records were also removed.

A new `Profit_per_Hectare` column was calculated to compare profitability across farms of different sizes.


```python
# 6. Data Cleaning

before_rows = len(df)

# Remove exact duplicate rows
df = df.drop_duplicates().copy()

# Fill numerical missing values using median
numeric_cols = df.select_dtypes(include=np.number).columns

for col in numeric_cols:
    if df[col].isnull().any():
        df[col] = df[col].fillna(df[col].median())

# Fill categorical missing values using mode
categorical_cols = df.select_dtypes(include='object').columns

for col in categorical_cols:
    if df[col].isnull().any():
        df[col] = df[col].fillna(df[col].mode()[0])

# Calculate profit per hectare
df["Profit_per_Hectare"] = df["Profit_INR"] / df["Farm_Area_Hectares"]

print("Rows before cleaning:", before_rows)
print("Rows after cleaning:", len(df))
print("Duplicate rows remaining:", df.duplicated().sum())
print("Missing values remaining:", int(df.isnull().sum().sum()))
```

```python
# 7. Verify Cleaned Data

print("Missing values after cleaning:")
print(df.isnull().sum().sum())

print("\nDuplicate rows after cleaning:")
print(df.duplicated().sum())

print("\nCleaned Dataset Shape:")
print(df.shape)
```

# Exploratory Data Analysis

The data was explored to understand the distribution of farms across seasons, crops, and states.

Different comparisons and visualizations were used to study seasonal performance, crop performance, environmental conditions, resource usage, and regional differences.

```python
# 9. Crop Distribution

crop_counts = df["Crop"].value_counts()

display(crop_counts)

plt.figure(figsize=(10, 5))
sns.countplot(data=df, x="Crop")

plt.title("Distribution of Farms Across Crops")
plt.xlabel("Crop")
plt.ylabel("Number of Farms")
plt.xticks(rotation=45)

plt.show()
```

```python
# 10. State Distribution

state_counts = df["State"].value_counts()

display(state_counts)

plt.figure(figsize=(10, 5))
sns.countplot(data=df, x="State")

plt.title("Distribution of Farms Across States")
plt.xlabel("State")
plt.ylabel("Number of Farms")
plt.xticks(rotation=45)

plt.show()
```

```python
# 12. Total Production by Season

season_production = df.groupby("Season")["Production_Tonnes"].sum().sort_values(ascending=False)

display(season_production)

plt.figure(figsize=(8, 5))
sns.barplot(x=season_production.index, y=season_production.values)

plt.title("Total Production by Season")
plt.xlabel("Season")
plt.ylabel("Total Production (Tonnes)")

plt.show()
```

```python
# 13. Total Revenue by Season

season_revenue = df.groupby("Season")["Revenue_INR"].sum().sort_values(ascending=False)

display(season_revenue)

plt.figure(figsize=(8, 5))
sns.barplot(x=season_revenue.index, y=season_revenue.values)

plt.title("Total Revenue by Season")
plt.xlabel("Season")
plt.ylabel("Revenue (INR)")

plt.show()
```
